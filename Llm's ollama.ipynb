```python
import requests
import json
# Define the endpoint URL
url = "http://localhost:11434/api/generate"

# Set up the request body (Change 'llama3' to whatever model you have installed)
payload = {
    "model": "deepseek-r1:8b",
    "prompt": "Why is the sky blue?",
    "stream": False  # Set to True if you want a chunked streaming response
}
```

```python
import subprocess
def stop_ollama(model_name):
    try:
        subprocess.Popen(
            f"ollama stop {model_name}",
            shell=True
        )
        print(f"Stopped ollama Successfully")
    except Exception as e:
        print(f"Error: {e}")
```

```python
law_system_prompt = """you are a ai assitant related Indian Laws, your task is to give a good verified 
                        information about anything related to indian law or realted cases, don't 
                        provide any unverified information, if no information is provided give it as no 
                        furture information is available or something like that, max no of tokens should be 1000, 
                        and 2-3 paragraphs, make sure to complete the sentance by end"""
payload2 = {
    "model": "deepseek-r1:8b",
    "prompt": law_system_prompt,
    "stream": False  # Set to True if you want a chunked streaming response
}

response2 = requests.post(url,json=payload2)
```

```python
if response2.status_code == 200:
    result = response2.json()
    print(result.get("response"))
else:
    print(f"Error: {response2.status_code} - {response2.text}")
```

```python
ollama_url = 'http://localhost:11434/v1'
MODEL = 'llama3.2:latest'
# MODEL = 'hf.co/varma007ut/Indian_Legal_Assitant:latest'
message = [{'role':'user', 'content':'''what is R^2 matric in ML, give the range ,
           and how to asses a model based on the metric (which rang of values are good and bad)'''}]
```

```python
import ollama

stream = ollama.chat(
        model = MODEL,
    messages=message,
    stream=True
)
```

```python
for chunk in stream:
    print(chunk['message']['content'],end='',flush=True)
stop_ollama(MODEL)
```

```python
law_system_prompt = """you are a ai assitant related Indian Laws, your task is to give a good verified 
                        information about anything related to indian law or realted cases, don't 
                        provide any unverified information, if no information is provided give it as no 
                        furture information is available or something like that, max no of tokens should be 1000, 
                        and 2-3 paragraphs, make sure to complete the sentance by end"""
stream = ollama.chat(
    # model='hf.co/varma007ut/Indian_Legal_Assitant:latest',
    model = "deepseek-r1:8b",
    messages=[{'role':'system', 'content': law_system_prompt},{'role': 'user', 'content':'What are the fundamental rights in the Indian Constitution?'}],
    stream=True,
    options={
        'num_predict': 1000  # Sets the maximum length to 150 tokens
    }
)
for chunk in stream:
    print(chunk['message']['content'],end='',flush=True)
```

```python
stop_ollama("deepseek-r1:8b")  # Stop the model after the request
```
