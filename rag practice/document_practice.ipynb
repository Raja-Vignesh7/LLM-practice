```python
from langchain_community.document_loaders import JSONLoader, PyMuPDFLoader, DirectoryLoader
from langchain_core.documents import Document

from datetime import datetime
import os
```

```python
doc = Document(
    page_content="This is the main text content I am using to create RAG",
    metadata={
        "source":"example.txt",
        "pages":1,
        "author":"Vignesh",
        "date_created": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
)
doc
```

```python
sample_pdf_path = r"inputs/pdf/polymers-13-01898.pdf"
pdf_loader = PyMuPDFLoader(sample_pdf_path)
pdf_doc = pdf_loader.load()
len(pdf_doc)
```

```python
dir_path = r"inputs/pdf"
dir_loader = DirectoryLoader(
    dir_path,
    glob="**/*.pdf",
    loader_cls= PyMuPDFLoader,
    show_progress=True
)

pdf_docs = dir_loader.load()
print(f"pdf_doc length: {len(pdf_docs)}")
print(f"No of pdf files in {dir_path}: {len(os.listdir(dir_path))}")
```
